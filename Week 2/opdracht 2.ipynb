```python

import numpy as np
import matplotlib.pyplot as plt
```

```python
def pixelpositie(pixel_x: int, pixel_y: int, width: int) -> complex:
    """deze functie bepaald het complexe getal voor een pixel positie
    gemaakt door Sven"""
    stapgrootte = 2 / width
    x = -1.5 + stapgrootte * pixel_x 
    y = -1 + stapgrootte * pixel_y
    c = complex(x, y)
    return c 

    
```

```python
def diverging(complex_getal: complex) -> int:
    """Bepaalt de diverging index van een complex getal.
    gemaakt door Sven"""
    a = 0
    n = 0
    for i in range(100):
        n+=1
        a = a**2 + complex_getal
        if abs(a) > 2:
            return n
    return 0
```

```python
def create_image(width: int) -> np.ndarray:
    """deze functie maakt een numpy array met alle kleur codes erin
    gemaakt door Sven"""
    image = np.zeros((width, width))
    for i in range(width):
        for j in range(width):
            positie = pixelpositie(j,i,width)
            code = diverging(positie)
            image[i][j] = code
    return image

```

```python
def draw_mandel(lengte: int) -> None:
    """hoofd functie die het plaatje plot
    gemaakt door Sven"""
    plaatje = create_image(lengte)
    plt.imshow(plaatje, extent = [-1.5, 0.5, -1, 1])
    plt.show()
    
```

```python
draw_mandel(500)
```
