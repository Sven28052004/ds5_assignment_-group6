```python
def pixelpositie(pixel_x, pixel_y, witdh):
    """deze functie bepaald het complexe getal voor een pixel positie
    gemaakt door Sven"""
    stapgrootte = 2 / witdh
    x = -1.5 + stapgrootte * pixel_x 
    y = -1 + stapgrootte * pixel_y
    c = complex(x, y)
    return c 

    
```

```python
def diverging(complex_getal):
    """Bepaalt de diverging index van een complex getal.
    gemaakt door Sven"""
    a = 0
    n = 0
    for i in range(100):
        n+=1
        a = a**2 + complex_getal
        if abs(a) > 2:
            return n
    return 0
```
